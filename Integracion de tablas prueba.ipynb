```python
# Cargar las dos tablas de OTUs en formato Excel
otu_table1 = pd.read_excel('otu_combined_anguil.xlsx', index_col=0)  # Ajusta el nombre si es necesario
otu_table2 = pd.read_excel('otu_combined_pergamino.xlsx', index_col=0)
```

```python
# Combinar ambas tablas basadas en las OTUs (índice)
otu_combined = pd.concat([otu_table1, otu_table2], axis=1, join='outer')
```

```python
# Guardar la tabla combinada en un nuevo archivo
otu_combined.to_csv('otu_combined.csv')
```

```python
# Cargar las tablas de taxonomía en formato Excel
taxonomy1 = pd.read_excel('taxonomy_combined_anguil.xlsx', index_col=0)  # Ajusta el nombre si es necesario
taxonomy2 = pd.read_excel('taxonomy_combined_pergamino.xlsx', index_col=0)
```

```python
# Cargar la tabla combinada de OTUs
otu_combined = pd.read_csv('otu_combined.csv', index_col=0)
```

```python
# Combinar las tablas de taxonomía basadas en las OTUs
taxonomy_combined = pd.concat([taxonomy1, taxonomy2], axis=0, join='outer')
```

```python
# Verificar si hay OTUs duplicadas en el índice
duplicated_otus = taxonomy_combined.index[taxonomy_combined.index.duplicated()]
print(duplicated_otus)
```

```python
# Eliminar filas duplicadas manteniendo la primera aparición
taxonomy_combined = taxonomy_combined[~taxonomy_combined.index.duplicated(keep='first')]
```

```python
# Reordenar la tabla de taxonomía según el orden de OTUs en la tabla combinada
taxonomy_combined = taxonomy_combined.reindex(otu_combined.index)
```

```python
# Guardar la tabla de taxonomía combinada en un nuevo archivo CSV
taxonomy_combined.to_csv('taxonomy_combined.csv')
```
